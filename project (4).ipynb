```python
# EMOTION LOGIC
def detect_emotion(text):
    txt = text.lower()
    if "sad" in txt or "depressed" in txt or "alone" in txt:
        return "SAD"
    elif "happy" in txt or "good" in txt or "excited" in txt:
        return "HAPPY"
    elif "angry" in txt or "mad" in txt:
        return "ANGRY"
    elif "fear" in txt or "scared" in txt:
        return "FEAR"
    else:
        return "NEUTRAL"

print("Emotion system ready! ✅")
```

```python
# CHATTING
print("Chatting is now ALLOWED! Start talking to Enzo...")
print("-" * 40)

while True:
    user_input = input("\nYou (type 'quit' to stop): ")

    if user_input.lower() == "quit":
        print("Enzo: Goodbye Maheen! Take care! ")
        break

    emotion = detect_emotion(user_input)

    if emotion == "SAD":
        print(f"Enzo [{emotion}]: I am here to support you. Don't worry, you are not alone.")
    elif emotion == "HAPPY":
        print(f"Enzo [{emotion}]: Wonderful! You seem happy! ")
    elif emotion == "ANGRY":
        print(f"Enzo [{emotion}]: Let's take a deep breath together. It's okay.")
    elif emotion == "FEAR":
        print(f"Enzo [{emotion}]: You are safe. I am with you.")
    else:
        print(f"Enzo [{emotion}]: I got your message - '{user_input}'. I am listening carefully.")
```
