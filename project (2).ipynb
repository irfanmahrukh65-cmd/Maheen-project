```python
from transformers import pipeline

# Model load ho raha hai
print("Model load ho raha hai, thoda wait karo...")
emotion_model = pipeline("text-classification", model="j-hartmann/emotion-english-distilroberta-base")
print("Model ready!")

# Aapka PIN
PIN_CODE = "2026"

# Step 1: PIN Verification (Sir isi se kholenge)
pin_input = input("PIN Daalo (2026): ")

if pin_input == PIN_CODE:
    print("\n PIN Correct! Welcome Maheen ")
    print("Private Chatbot Start Ho Gaya\n")

    # Step 2: Chat Loop
    while True:
        msg = input("aap: ")

        if msg.lower() == "exit":
            print("Enzo: Bye Maheen, milte hain ")
            break

        # Emotion Detect
        result = emotion_model(msg)[0]
        emotion_label = result['label']
        score = round(result['score'], 2)

        print(f"[Emotion Detected: {emotion_label} ({score})]")

        # Enzo ka Reply
        if emotion_label == 'sadness':
            print("Enzo: Udas mat ho, main yahan hun tumhare liye.")
        elif emotion_label == 'joy':
            print("Enzo: Wah, khush ho ke acha laga!")
        elif emotion_label == 'anger':
            print("Enzo: Gussa thook do, relax karo.")
        elif "kalma" in msg.lower():
            print("Enzo: Kalma chowk par rush ka khayal rakhna.")
        else:
            print(f"Enzo: Samajh gaya Maheen, tumne kaha '{msg}'")

else:
    print(" PIN Ghalat Hai, Access Denied!")
```

```python
from IPython.display import display, HTML
import ipywidgets as widgets

# --- FRONTEND ---
display(HTML("<h2 style='color:green; text-align:center;'>🔒 Maheen's Project Ready! PIN: 2026 - Frontend Working!</h2>"))

chat_box = widgets.Output(layout={'border': '1px solid #ccc', 'height':'300px', 'overflow_y':'scroll', 'padding':'10px'})

user_input = widgets.Text(
    placeholder='Type your message in English...',
    description='You:',
    layout=widgets.Layout(width='70%')
)
send_btn = widgets.Button(description="Send", button_style='success')

def on_send_clicked(b):
    with chat_box:
        text = user_input.value
        if text.strip() == "":
            return
        print(f"You: {text}")

        # --- BACKEND LOGIC - PURE ENGLISH ---
        txt = text.lower()
        if "sad" in txt or "depressed" in txt or "upset" in txt:
            reply = "Enzo: I detect you are feeling SAD. I am here for you."
        elif "happy" in txt or "excited" in txt or "good" in txt:
            reply = "Enzo: You seem HAPPY! That's wonderful!"
        elif "angry" in txt or "mad" in txt:
            reply = "Enzo: I sense ANGER. Let's calm down together."
        elif "fear" in txt or "scared" in txt:
            reply = "Enzo: It seems you are AFRAID. You are safe."
        else:
            reply = "Enzo: Welcome! Emotion Detection Active! Your emotion is NEUTRAL."

        print(reply + "\n---")
        user_input.value = ""

send_btn.on_click(on_send_clicked)
user_input.on_submit(lambda x: on_send_clicked(None))

display(chat_box)
display(widgets.HBox([user_input, send_btn]))
```
